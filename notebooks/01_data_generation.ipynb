```python
# Import Libraries
import pandas as pd
import random
from faker import Faker
from datetime import datetime, timedelta

fake = Faker()
Faker.seed(42)
random.seed(42)

print("Libraries imported successfully!")
```

```python
# Generate Mock E-Commerce Data
num_orders = 1500
categories = {
    'Electronics': [299.99, 599.99, 1099.99, 49.99],
    'Clothing': [19.99, 39.99, 89.99, 14.99],
    'Home & Kitchen': [24.99, 85.00, 150.00, 12.99],
    'Beauty': [9.99, 29.99, 75.00, 18.50]
}
genders = ['Male', 'Female', 'Non-binary']
cities = ['New York', 'Los Angeles', 'Chicago', 'Houston', 'Miami', 'Seattle']

data = []

start_date = datetime(2025, 1, 1)

for i in range(num_orders):
    order_id = f"ORD{10000 + i}"
    # Generate realistic dates across the year
    order_date = start_date + timedelta(days=random.randint(0, 365))
    customer_age = random.randint(18, 65)
    gender = random.choices(genders, weights=[45, 50, 5], k=1)[0]
    location = random.choice(cities)
    
    category = random.choice(list(categories.keys()))
    unit_price = random.choice(categories[category])
    quantity = random.randint(1, 5)
    
    data.append([
        order_id, order_date.strftime('%Y-%m-%d'), customer_age, 
        gender, location, category, quantity, unit_price
    ])

# Create DataFrame
df = pd.DataFrame(data, columns=[
    'Order_ID', 'Order_Date', 'Customer_Age', 
    'Gender', 'Location', 'Product_Category', 'Quantity', 'Unit_Price'
])

# Save to CSV
df.to_csv('C:\\ecommerce-insights-pipeline\\data\\raw_synced_data.csv', index=False)
print("Successfully generated raw_synced_data.csv in the data/ folder!")
```
